```python
from hazm import *
import json
```

# reading file

```python
f = open('IR_data_news_12k.json')
data = json.load(f)
```

# preprocessing : normalizing

```python
normalizer = Normalizer()
normal_content_list = []
```

```python
for i in data:
    normal_content = normalizer.normalize(data[i]['content'])
    normal_content_list.append(normal_content)
    print(normal_list)
```

# preprocessing : tokenizing

```python
word_tokenize_list = []
for normal_content in normal_content_list:
    word_tokenize_list.append(word_tokenize(normal_content))
```

```python
print(len(word_tokenize_list))
for i in range(10):
    print(word_tokenize_list[i])
```

# preprocessing : Lemmatizering

```python
lemmatizer = Lemmatizer()
```

```python
word_lem_list = []

for word_list in word_tokenize_list:
    temp_list = []
    for word in word_list:
        
        word = lemmatizer.lemmatize(word)
        
        temp_list.append(word)
        
    word_lem_list.append(temp_list)
```

```python
print(word_lem_list[0])
```

# preprocessing : Stemming --> ! issue: خیلی لغات رو خراب میکنه !لازمه اصن ؟

```python
stemmer = Stemmer()
```

```python
word_stem_list = []

for word_list in word_tokenize_list:
    temp_list = []
    for word in word_list:
        
        word = stemmer.stem(word)        
        
        temp_list.append(word)
        
    word_stem_list.append(temp_list)
```

```python
print(word_stem_list[0])
```
